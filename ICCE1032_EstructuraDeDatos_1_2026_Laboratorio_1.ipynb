## 5. Completa la clase

Ahora agrega métodos simples a la clase.

### Tareas
- Crear un método `mostrar_info()` que imprima la información principal de la sala.
- Crear un método `reservar()` que deje la sala como no disponible si estaba libre.
- Crear un método `liberar()` que deje la sala disponible nuevamente.

### Pista
Recuerda que dentro de los métodos puedes acceder a los atributos usando `self`.


```python
class Sala:
    def __init__(self, nombre, capacidad, ubicacion, disponible=True):
        self.nombre = nombre
        self.capacidad = capacidad
        self.ubicacion = ubicacion
        self.disponible = disponible

    # Completa este método
    def mostrar_info(self):
        estado = "Disponible" if self.disponible else "No disponible"
        print(f"Sala:      {self.nombre}")
        print(f"Capacidad: {self.capacidad} personas")
        print(f"Ubicación: {self.ubicacion}")
        print(f"Estado:    {estado}")

    # Completa este método
    def reservar(self):
        if self.disponible:
            self.disponible = False
            print(f"{self.nombre} ha sido reservada.")
        else:
            print(f"{self.nombre} ya está ocupada.")

    # Completa este método
    def liberar(self):
        if not self.disponible:
            self.disponible = True
            print(f"{self.nombre} ahora está disponible.")
        else:
            print(f"{self.nombre} ya estaba libre.")
```

## 9. Desafío opcional

Si terminas antes, intenta una o más de estas mejoras:

- agregar el atributo `tipo_sala`
- validar que la capacidad no sea negativa
- mostrar la información en un formato más claro
- crear una lista con varias salas y recorrerla para imprimir sus datos
- impedir que una sala ya reservada se vuelva a reservar sin aviso


```python
# Espacio para desafío opcional
class Sala:
    def __init__(self, nombre, capacidad, ubicacion, tipo_sala="Aula", disponible=True):
        self.nombre    = nombre
        self.tipo_sala = tipo_sala
        self.ubicacion = ubicacion
        self.disponible = disponible

        if capacidad < 0:
            print(f"Advertencia: capacidad inválida. Se asigna 0.")
            self.capacidad = 0
        else:
            self.capacidad = capacidad

    def mostrar_info(self):
        estado = " Disponible" if self.disponible else "✗ Ocupada"
        print(f"  {self.nombre} ({self.tipo_sala})")
        print(f"  Capacidad : {self.capacidad} personas")
        print(f"  Ubicación : {self.ubicacion}")
        print(f"  Estado    : {estado}")

    def reservar(self):
        if self.disponible:
            self.disponible = False
            print(f" {self.nombre} reservada correctamente.")
        else:
            print(f" {self.nombre} ya está ocupada.")

    def liberar(self):
        if not self.disponible:
            self.disponible = True
            print(f" {self.nombre} liberada correctamente.")
        else:
            print(f" {self.nombre} ya estaba libre.")



sala_a101 = Sala("A101",              35,  "Edificio A", "Aula")
lab_info  = Sala("Lab Informática 2", 25,  "Edificio B", "Laboratorio")
auditorio = Sala("Auditorio Central", 120, "Edificio C", "Auditorio")
sala_mala = Sala("Sala X",            -5,  "Edificio D")


salas = [sala_a101, lab_info, auditorio, sala_mala]

print(" Todas las salas \n")
for sala in salas:
    sala.mostrar_info()
```
